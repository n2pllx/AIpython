```python
#1부터 100까지 합을 출력하시오.
total=0
for i in range(1,101):
  total+=i
print(total)
```

```python
#1부터 100까지의 짝수의 합을 range()의 증가값을 이용하여 구하기
num=0
for a in range(2,101,2):
  num+=a
print(num)
```

```python
#정수 하나를 입력받아 그 수의 구구단을 '7x1=7' 형식으로 출력하기
gugudan=int(input("숫자를 입력하세요: "))
for n in range(1,10):
  print(f'{gugudan}x{n}={gugudan*n}')
```

```python
#2단부터 9단까지 구구단을 '7x1=7' 형식으로 출력하기
for m in range(2,10):
  for x in range(1,10):
    print(f'{m}x{x}={m*x}')
  print("="*6)
```

```python
#1부터 100까지 합을 출력하시오.
sum=0
y=1
while y<=100:
  sum+=y
  y+=1
print(sum)
```

```python
sum=0
y=1
while True:
  sum+=y
  y+=1
  if y>100:
    break
print(sum)
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하기
sum=0
while True:
  n=int(input("양의 정수를 입력하셰요: "))
  if n==0:
    break
  sum+=n
print(sum)
```

```python
#1부터 20까지의 수 중 짝수만 담은 리스트를 만들어 출력하기
evens=[]
for i in range(1,21):
  if i%2==0:
    evens.append(i)
print(evens)
evens.pop(7)
print(evens)
del evens[5]
print(evens)
evens.remove(14)
print(evens)
print(evens[-1])
```

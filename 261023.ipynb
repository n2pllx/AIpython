```python
countries_capitals = {
    "대한민국": "서울",
    "일본": "도쿄",
    "중국": "베이징",
    "미국": "워싱턴 D.C.",
    "영국": "런던",
    "프랑스": "파리",
    "독일": "베를린",
    "이탈리아": "로마",
    "캐나다": "오타와",
    "호주": "캔버라",
    "브라질": "브라질리아",
    "아르헨티나": "부에노스아이레스",
    "인도": "뉴델리",
    "베트남": "하노이",
    "태국": "방콕",
    "스페인": "마드리드",
    "러시아": "모스크바",
    "이집트": "카이로",
    "멕시코": "멕시코시티",
    "사우디아라비아": "리야드"
}
```

```python
#국가명을 입력하면 수도가 출력되는 프로그램 만들기
while True:
    country_name = input("국가 이름을 입력하세요: ")
    if country_name in countries_capitals:
        print(f"{country_name}의 수도는 {countries_capitals[country_name]}입니다.")
        break
    else:
        print("해당 국가를 찾을 수 없습니다.")
```
